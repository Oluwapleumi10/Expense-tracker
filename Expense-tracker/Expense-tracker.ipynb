```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
statement_df = pd.DataFrame(
    { "Date" : [],
     "Item" : [],
     "Amount" : [],
     "Type" : [] } )

statement_df
     
```

```python
new_entry = pd.DataFrame(
    {"Date" : ["11th, January"],
     "Item" : ["Food"],
     "Amount" : [3000],
     "Type" : ["Expense"]}
)
statement_df = pd.concat([statement_df,new_entry] , ignore_index=True)
```

```python
statement_df
```

```python
def add_transaction(current_df ,date,item,amount,types):
    new_entry = pd.DataFrame(
        {"Date" : [date],
         "Item" : [item],
         "Amount" : [amount],
         "Type" : [types] } )
    current_df = pd.concat([current_df,new_entry] , ignore_index=True)
    return current_df

```

```python
df = add_transaction(statement_df,"Today","pizza",5000,"Expenses")
df
```
